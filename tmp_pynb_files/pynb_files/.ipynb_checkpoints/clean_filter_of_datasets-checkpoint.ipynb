```python
import numpy as np
import matplotlib as plt
import pandas as pd
```

```python
df_owid = pd.read_csv("../tmpdatasets/owid-covid-data.csv")
```

```python
whd = pd.read_csv("../tmpdatasets/WHO-COVID-19-global-daily-data.csv")
```

```python
# Step 1: Rename WHO columns
whd = whd.rename(columns={
    'Date_reported': 'date',
    'Country': 'location',
    'Country_code': 'iso_code',
    'New_cases': 'new_cases',
    'Cumulative_cases': 'total_cases',
    'New_deaths': 'new_deaths',
    'Cumulative_deaths': 'total_deaths'
})

# Step 2: Now select
who_sel = whd[['iso_code', 'location', 'date', 'new_cases', 'total_cases', 'new_deaths', 'total_deaths']]
```

```python
owid_sel = df_owid[['iso_code', 'location', 'date', 'new_cases', 'total_cases', 'new_deaths', 'total_deaths']].copy()
who_sel = whd[['iso_code', 'location', 'date', 'new_cases', 'total_cases', 'new_deaths', 'total_deaths']].copy()
```

```python
# Unique country/location names
owid_countries = set(owid_sel['location'].unique())
who_countries = set(who_sel['location'].unique())

# Show mismatches (countries in one but not the other)
only_in_owid = owid_countries - who_countries
only_in_who = who_countries - owid_countries

print("🔴 Countries in OWID but not in WHO:", sorted(only_in_owid))
print("🔵 Countries in WHO but not in OWID:", sorted(only_in_who))

# Check ISO code mismatches
owid_iso = set(owid_sel['iso_code'].dropna().unique())
who_iso = set(who_sel['iso_code'].dropna().unique())

only_in_owid_iso = owid_iso - who_iso
only_in_who_iso = who_iso - owid_iso

print("🔴 ISO codes in OWID but not in WHO:", sorted(only_in_owid_iso))
print("🔵 ISO codes in WHO but not in OWID:", sorted(only_in_who_iso))
```

```python
import pycountry
import pandas as pd

def iso3_to_iso2(iso3):
    try:
        country = pycountry.countries.get(alpha_3=iso3)
        if country:
            return country.alpha_2
        else:
            return None
    except Exception:
        return None

# Example: Assuming owid_sel is your OWID dataframe with 'iso_code' column in ISO3 format
# Create a new column with ISO2 codes converted from ISO3

owid_sel['iso_code_2'] = owid_sel['iso_code'].apply(lambda x: iso3_to_iso2(x) if pd.notna(x) else None)

# Now owid_sel['iso_code_2'] has ISO2 codes matching WHO's iso_code

# You can check mismatches again
owid_iso2 = set(owid_sel['iso_code_2'].dropna().unique())
who_iso = set(who_sel['iso_code'].dropna().unique())

only_in_owid_iso2 = owid_iso2 - who_iso
only_in_who_iso = who_iso - owid_iso2

print("🔴 ISO2 codes in OWID but not in WHO:", sorted(only_in_owid_iso2))
print("🔵 ISO2 codes in WHO but not in OWID:", sorted(only_in_who_iso))
```
